### Import Python Libraries
For problem 1, we will need a number of python libraries, ``pandas`` allows us to load, clean and manipulate our data. ``numpy`` handles mathematical calculations on large sets of data. ``matplotlib.pyplot`` allows us to create charts, histograms, scatter plots and more visuals. And ``sklearn.datasets`` provides ready-made datasets such as the iris dataset.

```python
# Import relevant libraries
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
# Load the iris dataset
iris = load_iris(as_frame=True)
df = iris.frame
```

```python
# Have a look at the first few rows of the dataset
df.head()
```

There are 5 fields within our data frame. Target and it's values (0, 1, 2) represent the various Iris species.

- 0 = setosa
- 1 = versicolor
- 2 = virginica

```python
# Map species names to target values and add 'species name' column
species_array = iris.target_names[iris.target]
df['species_name'] = species_array
df.head()
```

```python
# Get the summary statistics of the dataset
df.describe()
```

```python
df.info()
```

### Histogram of all iris flowers
I need to extract the petal length for each unique target name automatically before I can plot the petal lengths of each species.

```python
# Using a for loop to run through each unique species name
species_data = [df[df['species_name'] == name]['petal length (cm)'] for name in iris.target_names]
plt.hist(species_data, bins = 10, stacked=True, label=iris.target_names)

plt.title('Petal Length by Iris Species')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Number of Flowers')
plt.legend()

plt.show()
```
